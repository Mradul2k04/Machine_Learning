## LINEAR ALGEBRA

Creating Vectors and Matrices

```python
import numpy as np
```

```python
# Creating 1D Array and check the shape
v = np.array([1, 2, 3])
print(v.shape)
```

```python
#Column vector (2D array)
v_col = np.array([[1], [2], [3]])
print(v_col.shape)
```

```python
print(v_col)
```

```python
# Matrix (2D array)
A = np.array([[1, 2, 3],
            [4, 5, 6]])
print(A.shape)
```

```python
print(A)
```

# VECTOR & MATRIX ADDITION

```python
# Vector addition
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
v_sum = v1 + v2
print(v_sum)
```

```python
# Matrix addition
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A + B
print(C)
```

# SCALAR MULTIPLICATION


```python
# Scalar multiplication
v = np.array([1, 2, 3])
c = 2
scaled = c * v
print(scaled)
```

```python
# Matrix scaling
A = np.array([[1, 2], [3, 4]])
scaled_A = 0.5 * A
print(scaled_A)
```

# MATRIX-VECTOR MULTIPLICATION

```python
# Basic matrix-vector multiplication
A = np.array([[1, 2, 3],
              [4, 5, 6]])
v = np.array([[7], [8], [9]])

result = A @ v
print(result)
```

```python
#Dot Product
result = np.dot(A, v)
print(result)
```

#  MATRIX-MATRIX MULTIPLICATION

```python
A = np.array([[1, 2],      
            [3, 4],
            [5, 6]])

B = np.array([[7, 8, 9],   
            [10, 11, 12]])

C = A @ B 
print(C)
```

```python
C = np.dot(A, B)
print(C)
```

#  MATRIX TRANSPOSE

```python
A = np.array([[1, 2, 3],
            [4, 5, 6]])
print(A.shape)   
```

```python
A_T = A.T
print(A_T.shape)
```

```python
print(A_T)
```
