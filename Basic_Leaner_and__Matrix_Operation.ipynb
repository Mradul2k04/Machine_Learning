```python
#Finding the Per month income of the Applicant
Montly_household_income=df['Applicant_Income']+df['Coapplicant_Income']
print("Montly income per Applicant+Coapplicant\n" ,Montly_household_income)
```

```python
#Monthly Amount Leaft After all Expense
Monthly_Savings=Montly_household_income-df['Monthly_Expense']
print("Monthly Savings\n" ,Monthly_Savings)
```

```python
#SCALAR MULTIPLICATION
#Growth Rate per Month
growth_rate = 1.05
projected_income = df.Applicant_Income * growth_rate
print("Projected Income\n" ,projected_income)
```

```python
print(f"\nGrowth amount:\n {(projected_income - df.Applicant_Income)}")
```

```python
#Matrix Transpose
matrix = df[[
    'Applicant_Income',
    'Coapplicant_Income',
    'Loan_Amount'
]].values
```

```python
transposed = matrix.T
print(transposed)
```

```python
# Dot Product
features = df[['CIBIL_Score', 'Applicant_Income', 'Employment_Length_Years']].values
weights = np.array([0.5, 0.3, 0.2])

risk_scores = features @ weights
print("Risk Scores\n" ,risk_scores)
```

```python
#Covariance & Correlation Matrix
fin_cols = ['Applicant_Income', 'Loan_Amount', 'CIBIL_Score', 'Debt_to_Income_Ratio', 'Asset_Value']
cov_matrix = np.cov(df[fin_cols].values.T)
print("Covariance Matrix\n" ,cov_matrix)
```

```python
corr_matrix = np.corrcoef(df[fin_cols].values.T)
print("Correlation Matrix\n" ,corr_matrix)
```

```python
#Scaler Vector Multtiplication
#Rising the Applicant Inome by 7%
df['Applicant_Income'] = df['Applicant_Income'] * 1.07
print(df['Applicant_Income'])
```

```python
#Rate of Interset By Bank
#Rate of intrest increase by bank 5%
df['Rate_of_Interest'] = df['Loan_Amount'] * 0.05
print(df['Rate_of_Interest'])
```
